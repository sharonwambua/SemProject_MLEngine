```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('Mumbai_property_details.csv')
df.head()
```

a.Number of rows and Columns

```python
rows, cols = df.shape
print(f"Number of Rows: {rows}")
print(f"Number of Columns: {cols}")
```

b. Datatypes

```python
df.info()
```

c. Dataset completion

```python
missing_values_percentage = df.isnull().sum() * 100 / len(df)
missing_info = pd.DataFrame({
    'Missing Count': df.isnull().sum(),
    'Missing Percentage': missing_values_percentage
})
missing_info = missing_info.sort_values(by='Missing Count', ascending=False)
display(missing_info[missing_info['Missing Count'] > 0][['Missing Count']])

if missing_info['Missing Count'].sum() == 0:
    print("No missing values found in the dataset.")
```

d.Removing first 15 rows and last 20 rows an merging

```python
first_15_rows = df.head(15)
last_20_rows = df.tail(20)
df_merged_sections = pd.concat([first_15_rows, last_20_rows])
print("first 15 rows:")
display(df_merged_sections.head(15))
print("last 20 rows:")
display(df_merged_sections.tail(20))
```
